```python
import pandas as pd
from sqlalchemy import create_engine, URL

conn_url = URL.create(
    "mssql+pyodbc", host=r"localhost\SQLEXPRESS",
    database="BA_Airline_Commercial_Analytics",
    query={"driver": "ODBC Driver 17 for SQL Server", "trusted_connection": "yes"},
)
engine = create_engine(conn_url)
print("connected")
```

```python
airport_names = pd.read_sql("""
    SELECT DISTINCT airport_1 AS code, city1 AS city_name FROM bronze.route_market_fares
    UNION
    SELECT DISTINCT airport_2 AS code, city2 AS city_name FROM bronze.route_market_fares
""", engine)
airport_names = airport_names.drop_duplicates(subset='code')
airport_names.to_parquet(r"..\app\data\airport_names.parquet")
print(f"Saved {len(airport_names)} airport name mappings")
```
